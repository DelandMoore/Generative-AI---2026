```python
#Question 4
query = np.array([3,4])
key = np.array([4,3])
total = sum(x*y for x,y in zip(query,key))
print(total)
print(query@key)
print(np.dot(query,key))


```

```python
fig, ax=plt.subplots(figsize=(6,4))


ax.set_xlim(-1, 6)
ax.set_ylim(-1,6)
plt.quiver(0,0, query[0], query[1], angles='xy', scale_units='xy', scale=1, color='navy', label='query = [3,4]')
plt.quiver(0,0, key[0], key[1], angles='xy', scale_units='xy', scale=1, color='red', label='key = [4,3]')
```

```python
abs_q = np.linalg.norm(query)
abs_k = np.linalg.norm(key)
S = abs_q*abs_k
print(S)
```

```python
cos_theta = total/S
cos_theta
```

```python
theta = np.arccos(cos_theta) 
theta
```

```python
degrees = np.degrees(theta)
degrees
```

```python
fig, ax=plt.subplots(figsize=(6,4))


ax.set_xlim(-1, 6)
ax.set_ylim(-1,6)
plt.quiver(0,0, query[0], query[1], angles='xy', scale_units='xy', scale=1, color='navy', label='query = [3,4]')
plt.quiver(0,0, key[0], key[1], angles='xy', scale_units='xy', scale=1, color='red', label='key = [4,3]')
ax.text(0.5, 5, f"angle = {degrees: .1f} degrees")
plt.show()
```
