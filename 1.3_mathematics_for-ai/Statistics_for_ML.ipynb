## Variance and Standard Deviation

```python
#Variance: avg square dist of each point from the mean

scores = np.array([60,70,80,90,100])
mean = np.mean(scores)
deviation = scores - mean  # vectorized: substracts mean from every elt
sq_deviations = deviation ** 2 #squares each deviation

variance = np.var(scores, ddof=1) #ddof=1 -> divide by (n-1), sample variance
std_dev = np.std(scores,ddof=1)

print("Variance: ", variance)
print("Std Deviation: ",std_dev.round(2))
```
